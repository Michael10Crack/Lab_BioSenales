### *Consultar historia, uso y proyectos futuros relacionados con Interfaz Cerebro Computador*

Las interfaces cerebro-computador (BCI, por sus siglas en inglés) son sistemas que permiten establecer una comunicación directa entre el cerebro y dispositivos externos sin la participación del sistema neuromuscular. Estas tecnologías se basan en la adquisición y procesamiento de señales cerebrales, comúnmente mediante electroencefalografía (EEG), con el fin de traducir la actividad neuronal en comandos que pueden ser utilizados para controlar sistemas electrónicos.

Los fundamentos de las BCI se remontan a 1924, cuando Hans Berger realizó el primer registro de un electroencefalograma humano, sentando las bases para el estudio de la actividad eléctrica cerebral. Posteriormente, en 1973, Jacques Vidal formalizó el concepto de BCI al proponer el uso de señales EEG para el control de dispositivos externos. Durante las décadas de 1980 y 1990, investigaciones como las lideradas por Wolpaw en el Instituto Wadsworth demostraron que los sujetos podían aprender a modular voluntariamente sus ritmos sensoriomotores para controlar interfaces, como el movimiento de un cursor en pantalla.

A partir de los años 2000, el campo de las BCI experimentó un crecimiento significativo impulsado por la disponibilidad de bases de datos públicas, la realización de competencias internacionales y el desarrollo de algoritmos de procesamiento y clasificación más robustos. Esto permitió mejorar la precisión y confiabilidad de estos sistemas, facilitando su transición desde entornos experimentales hacia aplicaciones más prácticas.

En la actualidad, las BCI tienen diversas aplicaciones, especialmente en el ámbito clínico y de la rehabilitación. Se emplean en el control de prótesis y dispositivos de asistencia para personas con discapacidades motoras, como pacientes con parálisis o enfermedades neurodegenerativas. Asimismo, permiten el desarrollo de sistemas de comunicación alternativa para individuos con limitaciones severas, y se utilizan en procesos de rehabilitación neurológica mediante técnicas de neuroretroalimentación. Además, han comenzado a explorarse en áreas como el control de dispositivos robóticos, videojuegos, realidad virtual y monitoreo cognitivo.

A futuro, se espera que las BCI evolucionen hacia sistemas más precisos y accesibles, integrando técnicas de inteligencia artificial para mejorar la interpretación de las señales cerebrales, así como el desarrollo de sistemas híbridos que combinen EEG con otras modalidades. Estas mejoras permitirán ampliar su aplicación en medicina, rehabilitación y diversas áreas de la bioingeniería.


### *Consultar sobre actividad cerebral registrada por EEG durante movimientos e imaginación de movimiento de las manos y cómo se hace su análisis en sistemas BCI: Electrodos, ritmos cerebrales de interés*

El **electroencefalograma** (EEG) es una técnica no invasiva que permite registrar la actividad eléctrica del cerebro mediante electrodos ubicados sobre el cuero cabelludo. Estas señales reflejan la actividad sincronizada de poblaciones neuronales y son ampliamente utilizadas en el estudio de procesos motores y cognitivos.

Durante la ejecución de movimientos voluntarios, se produce una activación de la corteza motora que se manifiesta como cambios en la potencia de ciertas bandas de frecuencia del EEG. En particular, los ritmos **mu (8–12 Hz)** y **beta (13–30 Hz)**, asociados a estados de reposo motor, presentan una *disminución de su potencia* cuando el sujeto realiza un movimiento. Este fenómeno se conoce como **desincronización relacionada con eventos** (ERD, Event-Related Desynchronization) y refleja un aumento en la actividad neuronal desorganizada asociada a la ejecución de tareas motoras. De manera complementaria, tras la finalización del movimiento puede observarse un incremento en la potencia de estas bandas, fenómeno conocido como **sincronización relacionada con eventos** (ERS, Event-Related Synchronization), asociado a procesos de recuperación o inhibición cortical.

En el caso de la imaginación de movimiento (motor imagery), aunque no existe ejecución física, se activan regiones similares de la corteza motora, generando patrones de ERD comparables a los del movimiento real, aunque generalmente de menor amplitud. Esta característica es fundamental en sistemas BCI, ya que permite inferir la intención de movimiento a partir de la actividad cerebral sin necesidad de acción muscular.

Desde el punto de vista espacial, estas variaciones se observan principalmente en electrodos ubicados sobre la corteza motora según el sistema internacional 10-20. Los electrodos C3, C4 y Cz son tradicionalmente reportados como los de mayor interés: C3 (hemisferio izquierdo) se asocia al control motor de la mano derecha, C4 (hemisferio derecho) al de la mano izquierda, y Cz a la actividad de la línea media motora. No obstante, distintos estudios han reportado contribuciones significativas de regiones parietales y centro-parietales en tareas de movimiento e imaginación motora, particularmente cuando intervienen procesos de planificación e integración sensoriomotora. Por esta razón, en este proyecto se calcula la PSD sobre todos los canales válidos y se identifican los canales discriminativos mediante un criterio empírico de varianza entre condiciones, contrastando posteriormente los hallazgos con los electrodos motores clásicos.

En cuanto a los ritmos cerebrales, el EEG se clasifica convencionalmente en las bandas delta (0.5–4 Hz), theta (4–8 Hz), alpha (8–12 Hz), beta (13–30 Hz) y gamma (>30 Hz). Es importante notar que el ritmo mu corresponde al ritmo alpha registrado específicamente sobre la corteza sensoriomotora, por lo que comparte su rango de frecuencia (8–12 Hz).

 En el contexto de sistemas BCI orientados al control motor, las bandas mu y beta son las más relevantes debido a su relación directa con la activación e inhibición de la corteza motora. La banda gamma, aunque de interés creciente en investigación, tiene menor uso en BCI clásicas debido a su mayor susceptibilidad al ruido muscular.


###  *Consultar y discutir los flujos de procesamiento propuestos para estos tipos de proyectos y que se puedan realizar en el marco del curso.* 

#### Flujos de procesamiento para señales EEG en sistemas BCI

En la literatura se han propuesto diversos flujos de procesamiento para el análisis de señales EEG en sistemas BCI basados en imaginación de movimiento. 

#### Flujos identificados en la literatura

**Flujo 1: Pipeline clásico basado en análisis espectral (PSD)**

Este es el enfoque más tradicional y ampliamente documentado en BCI motor. Consiste en preprocesamiento de la señal (filtrado pasa-banda, notch y re-referenciación), segmentación por eventos, estimación de la densidad espectral de potencia (PSD) mediante el método de Welch en las bandas de interés (mu y beta), y análisis estadístico comparativo entre condiciones. Su principal ventaja es la transparencia matemática y la interpretación fisiológica directa de los resultados en términos de ERD y ERS.

**Flujo 2: Pipeline basado en filtros espaciales (CSP + clasificador)**

El método de Patrones Espaciales Comunes (CSP, Common Spatial Patterns) es uno de los enfoques más utilizados en competencias BCI. Aplica un filtro espacial que maximiza la varianza de una clase mientras minimiza la de otra, extrayendo características discriminativas entre condiciones. Aunque reporta buena precisión de clasificación, requiere un número suficiente de ensayos por condición y no ofrece una interpretación espectral directa de la actividad cerebral.

**Flujo 3: Pipeline basado en transformadas tiempo-frecuencia (STFT / Wavelet)**

Este enfoque utiliza la Transformada de Fourier de Tiempo Corto (STFT) o la Transformada Wavelet para obtener representaciones tiempo-frecuencia de la señal, permitiendo observar cómo evoluciona la potencia espectral a lo largo del tiempo. Es útil cuando se requiere analizar la dinámica temporal de la ERD/ERS. Sin embargo, implica mayor complejidad computacional e interpretativa respecto al análisis espectral clásico.

**Flujo 4: Pipeline basado en aprendizaje profundo (Deep Learning)**

Los enfoques más recientes utilizan redes neuronales convolucionales (CNN) o modelos híbridos que reciben directamente la señal cruda o representaciones tiempo-frecuencia como entrada, eliminando la necesidad de extracción manual de características. Aunque reportan alta precisión, requieren grandes volúmenes de datos, recursos computacionales elevados y ofrecen baja interpretabilidad de los resultados. 

#### Flujo seleccionado para este proyecto

Para este proyecto se adopta el **Pipeline clásico basado en análisis espectral (PSD)**, por las siguientes razones:

En primer lugar, desde el punto de vista del curso de Bioseñales, cada etapa del flujo está directamente vinculada a los contenidos trabajados: el filtrado digital, el análisis espectral mediante la transformada de Fourier y las herramientas estadísticas sobre señales. 

En segundo lugar, desde el punto de vista fisiológico, la PSD es la herramienta más adecuada dado que los fenómenos de interés — la ERD y la ERS — se manifiestan precisamente como cambios de potencia en bandas específicas del EEG. 

En tercer lugar, a diferencia de los enfoques basados en CSP, transformadas tiempo-frecuencia o aprendizaje profundo, el pipeline espectral no requiere un número elevado de ensayos por sujeto, ni recursos computacionales especializados, ni conocimientos previos de aprendizaje automático, lo que lo hace viable y apropiado para el alcance del proyecto.

En teoría el flujo adoptado comprendería las siguientes etapas: preprocesamiento (filtrado pasa-banda 1–40 Hz, filtro notch, re-referenciación CAR y rechazo de épocas), segmentación por eventos, extracción de la PSD por bandas y canales de interés, y análisis estadístico comparativo entre las cuatro condiciones experimentales.

#### Preprocesamiento

Dado que la señal EEG cruda contiene múltiples fuentes de contaminación, el preprocesamiento busca mejorar la calidad de la señal sin introducir transformaciones complejas. Las etapas aplicadas son:

- **Filtrado pasa-banda (1–40 Hz):** Conserva las frecuencias relevantes del EEG, eliminando la deriva de línea base y el ruido de alta frecuencia no relacionado con los ritmos cerebrales de interés.

- **Filtro Notch (60 Hz):** Suprime la interferencia de la red eléctrica presente en los registros.

- **Re-referenciación por promedio común (CAR):** Mejora la relación señal-ruido espacial restando a cada canal el promedio de todos los canales en cada instante de tiempo.

- **Rechazo de épocas por umbral de amplitud:** Descarta segmentos contaminados por artefactos oculares o musculares de gran magnitud, conservando únicamente los segmentos de señal confiables.

#### Segmentación

Una vez preprocesada la señal continua, los registros se dividen en ventanas temporales llamadas épocas, alineadas a los eventos registrados en el archivo de anotaciones de cada sujeto. Cada época queda etiquetada con la condición experimental correspondiente:

- Movimiento real de mano derecha
- Movimiento real de mano izquierda  
- Imaginación de movimiento de mano derecha
- Imaginación de movimiento de mano izquierda

Esta segmentación es fundamental para comparar condiciones experimentales de manera consistente y aislar la respuesta cerebral asociada a cada tarea.

#### Extracción de características — Densidad Espectral de Potencia (PSD)

Una vez segmentadas las señales, se realiza la extracción de características mediante análisis espectral. La estimación de la **densidad espectral de potencia (PSD)** mediante el método de Welch permite cuantificar la distribución de energía en diferentes bandas de frecuencia del EEG. Este método divide cada época en segmentos con solapamiento, calcula la transformada de Fourier de cada uno y promedia los resultados, obteniendo una estimación estable de la potencia en función de la frecuencia.

La PSD se calcula para los canales de interés sobre la corteza motora (C3, C4, Cz y adyacentes) y se integra en cada banda de frecuencia:

| Banda     | Rango de frecuencia | Relevancia en BCI motor                        |
|-----------|---------------------|------------------------------------------------|
| Delta     | 1 – 4 Hz          | Baja relevancia, asociada a sueño profundo     |
| Theta     | 4 – 8 Hz            | Procesos cognitivos y memoria de trabajo       |
| Mu/Alpha  | 8 – 12 Hz           | **Alta relevancia:** ERD/ERS en corteza motora |
| Beta      | 13 – 30 Hz          | **Alta relevancia:** ERD/ERS en corteza motora |
| Gamma     | > 30 Hz             | Procesamiento de alto nivel, susceptible a EMG |

*Nota: el límite inferior de la banda delta se establece en 1 Hz en concordancia con el filtro pasa-banda aplicado en el preprocesamiento (1–40 Hz).*

#### Análisis estadístico

Con las características extraídas se procede al análisis comparativo entre las cuatro condiciones experimentales. El flujo incluye:

- 1. **Estadística descriptiva:** Visualización de la distribución de potencia por banda y condición mediante boxplots o violin plots, permitiendo identificar tendencias y diferencias entre hemisferios y condiciones.

- 2. **Prueba de normalidad (Shapiro-Wilk):** Determina si los datos siguen una distribución normal, lo cual define la elección de la prueba estadística.

- 3. **Pruebas de hipótesis:** Según el resultado de la prueba de normalidad, se aplican pruebas paramétricas o no paramétricas identificar si existen diferencias significativas.



#### Discusión del flujo de procesamiento

El flujo de procesamiento descrito corresponde a un enfoque ampliamente utilizado en sistemas BCI basados en EEG, debido a su equilibrio entre complejidad computacional y capacidad de extracción de información relevante.

El uso de técnicas de preprocesamiento como el filtrado pasa-banda y la eliminación de ruido permite mejorar la calidad de la señal sin introducir transformaciones complejas, lo cual resulta adecuado dentro del marco del curso de Bioseñales. Cada etapa está directamente vinculada a los contenidos trabajados: el filtrado digital, el análisis espectral mediante la transformada de Fourier y las pruebas estadísticas sobre señales.

La elección del análisis espectral mediante la PSD responde a su capacidad para caracterizar de manera eficiente la distribución de energía en las bandas de frecuencia de interés, especialmente en los ritmos mu y beta, los cuales están directamente relacionados con la actividad motora cortical. Aunque existen métodos más avanzados, como transformadas tiempo-frecuencia o técnicas basadas en aprendizaje profundo, estos requieren mayor complejidad computacional y una interpretación más elaborada, lo que podría excedre el alcance del curso.

Asimismo, la segmentación de la señal basada en eventos permite realizar comparaciones directas entre condiciones experimentales, facilitando el análisis estadístico posterior. 
Sin embargo, este enfoque depende de una correcta sincronización de los eventos y puede verse afectado por variabilidad entre sujetos.

1. Identificar empíricamente, mediante un criterio de varianza entre condiciones, cuáles canales presentan mayor capacidad discriminativa con los datos disponibles.
2. **Contrastar** los canales identificados estadísticamente con los electrodos motores clásicos (C3, C4, Cz) reportados en la literatura, verificando o refutando las predicciones teóricas.

Las bandas de mayor interés son **Mu y Beta**, por su asociación directa con los fenómenos de ERD/ERS en la corteza motora.

**Etapa 5 — Construcción del DataFrame de resultados**

Los resultados se almacenarán en un DataFrame estructurado donde cada fila corresponde a un registro de potencia por sujeto, condición y canal. Esta estructura permite realizar comparaciones sistemáticas entre condiciones, entre sujetos y entre canales de forma eficiente.

**Etapa 6 — Cálculo de ERD/ERS**

Sobre el DataFrame de potencias se calcula la **Desincronización/Sincronización Relacionada con Eventos (ERD/ERS)** como el cambio porcentual de potencia de cada condición de tarea respecto al reposo del mismo sujeto y canal:

$$ERD/ERS\,(\%) = \frac{P_{tarea} - P_{reposo}}{P_{reposo}} \times 100$$

Valores **negativos** indican ERD (activación cortical) y valores **positivos** indican ERS (sincronización/inhibición). Esta normalización permite comparar entre sujetos y canales reduciendo la variabilidad inter-individual de la potencia absoluta.

**Etapa 7 — Análisis estadístico**

El análisis estadístico se realiza sobre los valores de **ERD/ERS**, que es la variable de interés fisiológico, en dos niveles:

- **Estadística descriptiva:** boxplots de la distribución de potencia y de ERD/ERS por banda, canal y condición. Identificación de canales relevantes mediante varianza entre condiciones (top-10 en Mu y top-10 en Beta).

- **Pruebas inferenciales:** se evalúa primero la **normalidad** de los datos por grupo mediante la prueba de **Shapiro-Wilk**. Según el resultado se aplica el conjunto de pruebas paramétricas o no paramétricas correspondiente, junto con el análisis post-hoc cuando sea necesario.

#### 2. Hipótesis

Las hipótesis se plantean sobre los valores de **ERD/ERS** en las bandas **Mu y Beta**, ya que son las directamente asociadas al control motor cortical. La unidad de análisis es el sujeto (un valor por sujeto, condición y canal), de modo que las comparaciones entre condiciones son **pareadas** (mismo sujeto en distintas condiciones).

**Hipótesis 1 — Lateralización hemisférica durante movimiento real**

El principio de control motor contralateral predice que durante el movimiento real de una mano, el hemisferio contralateral presenta mayor desincronización (ERD más negativa) que el ipsilateral.

- **H₀:** No existe diferencia en el ERD/ERS de las bandas Mu y Beta entre los electrodos C3 y C4 durante el movimiento real de cada mano.
- **H₁:** Existe diferencia significativa en el ERD/ERS entre C3 y C4, siendo más negativo en el electrodo contralateral a la mano en movimiento.
- **Prueba:** comparación pareada entre C3 y C4 dentro de cada condición → **Wilcoxon pareado** (no paramétrica) o **t pareado** (paramétrica) según el resultado de Shapiro-Wilk.

**Hipótesis 2 — Diferencia entre movimiento real e imaginación**

El movimiento real activa más intensamente la corteza motora primaria que la imaginación, por lo que se espera mayor ERD durante movimiento real.

- **H₀:** No existe diferencia en el ERD/ERS de las bandas Mu y Beta entre las condiciones de movimiento real e imaginación de movimiento para una misma mano.
- **H₁:** El ERD/ERS es significativamente más negativo (mayor desincronización) durante el movimiento real que durante la imaginación.
- **Prueba:** comparación pareada por sujeto entre `mov_real_der` vs `img_der` y entre `mov_real_izq` vs `img_izq` → **Wilcoxon pareado** o **t pareado**.

**Hipótesis 3 — Diferencia global entre las cuatro condiciones experimentales**

- **H₀:** No existe diferencia en el ERD/ERS de las bandas Mu y Beta entre las cuatro condiciones experimentales.
- **H₁:** Al menos una de las cuatro condiciones presenta un ERD/ERS significativamente diferente a las demás.
- **Prueba:** comparación de ≥3 grupos relacionados → **Friedman** (no paramétrica) o **ANOVA de medidas repetidas** (paramétrica). Si se rechaza H₀, se aplica análisis **post-hoc**: **Wilcoxon pareado con corrección de Bonferroni** o **post-hoc de Dunn** según corresponda.

#### 3 Herramientas estadísticas

| Herramienta                 | Uso                                                                |
|-----------------------------|--------------------------------------------------------------------|
| Shapiro-Wilk                | Evaluar normalidad de los datos de ERD/ERS por grupo               |

El nivel de significancia adoptado para todas las pruebas es **α = 0.05**.

```python

# PASO 2: CARGAR Y CONCATENAR RUNS DE UN SUJETO


def cargar_sujeto(archivos, sujeto):
    """
    Carga y concatena todos los runs de interés de un sujeto,
    etiquetando correctamente cada condición.
    """
    runs_sujeto = sorted([a for a in archivos if sujeto in os.path.basename(a)])

    print(f"\nSujeto {sujeto}: {len(runs_sujeto)} runs encontrados")

    raws = []
    for ruta in runs_sujeto:
        nombre = os.path.basename(ruta)
        raw_temp = read_raw_edf(ruta, preload=True, verbose=False)
        mne.datasets.eegbci.standardize(raw_temp)
        raw_temp.set_montage('standard_1005')

        # Etiquetar según el run
        if any(r in nombre for r in ['R03', 'R07', 'R11']):
            # Movimiento real
            mapping = {'T0': 'reposo', 'T1': 'mov_real_izq', 'T2': 'mov_real_der'}
        else:
            # Imaginación de movimiento
            mapping = {'T0': 'reposo', 'T1': 'img_izq', 'T2': 'img_der'}

        raw_temp.annotations.rename(mapping)
        raws.append(raw_temp)

    raw = mne.concatenate_raws(raws, verbose=False)
    print(f"  Señal concatenada: {raw.times[-1]:.1f} segundos")

    # Eliminar anotaciones de boundary que interfieren con la segmentación
    raw.annotations.delete(
        [i for i, desc in enumerate(raw.annotations.description) 
            if 'boundary' in desc.lower()]
    )
    print(f"  Anotaciones después de limpiar: {len(raw.annotations)}")

    return raw
```

```python

# PASO 3: PREPROCESAMIENTO


def preprocesar_eeg(raw, visualizar=False):
    """
    Aplica el flujo de preprocesamiento sobre todos los canales.
    """
    raw_original = raw.copy()

    # FASE 1: Señal cruda
    print("=" * 50)
    print("FASE 1: Señal cruda")
    if visualizar:
        raw_original.plot(
            duration=30,
            title='Fase 1 — Señal cruda',
            scalings='auto'
        )
        plt.show()

    # FASE 2: Filtrado
    print("=" * 50)
    print("FASE 2: Filtrado pasa-banda (1–40 Hz) + Notch (60 Hz)")
    raw.filter(1, 40, verbose=False)
    raw.notch_filter(60, verbose=False)

    if visualizar:
        raw.plot(
            duration=30,
            title='Fase 2 — Después del filtrado',
            scalings='auto'
        )
        fig, axes = plt.subplots(1, 2, figsize=(12, 4))
        raw_original.compute_psd(fmax=60).plot(axes=axes[0], show=False)
        axes[0].set_title('PSD — Señal cruda')
        raw.compute_psd(fmax=60).plot(axes=axes[1], show=False)
        axes[1].set_title('PSD — Después del filtrado')
        plt.tight_layout()
        plt.show()

    # FASE 3: Re-referenciación CAR
    print("=" * 50)
    print("FASE 3: Re-referenciación CAR")
    raw.set_eeg_reference('average', verbose=False)
    raw.apply_proj()

    if visualizar:
        raw.plot(
            duration=30,
            title='Fase 3 — Re-referenciación CAR',
            scalings='auto'
        )
        plt.show()

    # FASE 4: Segmentación
    print("=" * 50)
    print("FASE 4: Segmentación basada en eventos")
    eventos, id_eventos = mne.events_from_annotations(raw, verbose=False)
    print(f"  Eventos encontrados: {len(eventos)}")
    print(f"  Condiciones: {id_eventos}")

    # FASE 5: Épocas — todos los canales EEG
    print("=" * 50)
    print("FASE 5: Épocas")
    epochs = mne.Epochs(
        raw,
        eventos,
        event_id=id_eventos,
        tmin=-1,
        tmax=4,
        picks='eeg',           # ← todos los canales
        baseline=(-1, 0),
        reject=dict(eeg=200e-6),
        preload=True,
        verbose=False
    )

    print(f"  Épocas totales: {len(epochs)}")
    for cond in epochs.event_id:
        print(f"    {cond}: {len(epochs[cond])}")

    if visualizar:
        epochs.plot(n_epochs=5, title='Fase 5 — Épocas segmentadas', scalings=dict(eeg=50e-6))
        plt.show()

    return epochs
```

### Función de preprocesamiento

La función `preprocesar_eeg` recibe una señal EEG cruda concatenada de un sujeto y aplica las cinco fases del pipeline de procesamiento definido. Cada fase está separada y, cuando el parámetro `visualizar=True`, se generan gráficas que permiten observar cómo evoluciona la señal en cada etapa del flujo.

**Las cinco fases del pipeline son:**

1. **Señal cruda:** se grafica la señal original sin ningún procesamiento, como referencia visual.
2. **Filtrado pasa-banda y notch:** se aplica un filtro pasa-banda de 1–40 Hz para conservar solo las frecuencias relevantes del EEG, y un filtro notch en 60 Hz para eliminar la interferencia de la red eléctrica. Se incluye una comparación de la PSD antes y después del filtrado.
3. **Re-referenciación CAR (Common Average Reference):** se sustrae el promedio de todos los canales en cada instante de tiempo, mejorando la relación señal-ruido espacial.
4. **Segmentación basada en eventos:** se identifican los marcadores de eventos en las anotaciones del archivo y se cuentan los eventos por condición.
5. **Creación de épocas:** la señal continua se divide en ventanas alineadas a cada evento (de -1 s a +4 s respecto al inicio del evento), excluyendo los canales contaminados identificados previamente.

El parámetro `visualizar` permite activar o desactivar las gráficas; se mantiene activo en la prueba con un sujeto y se desactiva cuando se procesan todos los sujetos para agilizar la ejecución.

```python
def preprocesar_eeg(raw, visualizar=False):
    """
    Aplica el flujo de preprocesamiento sobre todos los canales,
    excluyendo los canales frontales contaminados por artefactos oculares.
    """
    raw_original = raw.copy()

    # FASE 1: Señal cruda
    print("=" * 50)
    print("FASE 1: Señal cruda")
    if visualizar:
        raw_original.plot(
            duration=30,
            title='Fase 1 — Señal cruda',
            scalings='auto'
        )
        plt.show()

    # FASE 2: Filtrado
    print("=" * 50)
    print("FASE 2: Filtrado pasa-banda (1–40 Hz) + Notch (60 Hz)")
    raw.filter(1, 40, verbose=False)
    raw.notch_filter(60, verbose=False)

    if visualizar:
        raw.plot(
            duration=30,
            title='Fase 2 — Después del filtrado',
            scalings='auto'
        )
        fig, axes = plt.subplots(1, 2, figsize=(12, 4))
        raw_original.compute_psd(fmax=60).plot(axes=axes[0], show=False)
        axes[0].set_title('PSD — Señal cruda')
        raw.compute_psd(fmax=60).plot(axes=axes[1], show=False)
        axes[1].set_title('PSD — Después del filtrado')
        plt.tight_layout()
        plt.show()

    # FASE 3: Re-referenciación CAR
    print("=" * 50)
    print("FASE 3: Re-referenciación CAR")
    raw.set_eeg_reference('average', verbose=False)
    raw.apply_proj()

    if visualizar:
        raw.plot(
            duration=30,
            title='Fase 3 — Re-referenciación CAR',
            scalings='auto'
        )
        plt.show()

    # FASE 4: Segmentación
    print("=" * 50)
    print("FASE 4: Segmentación basada en eventos")
    eventos, id_eventos = mne.events_from_annotations(raw, verbose=False)
    print(f"  Eventos encontrados: {len(eventos)}")
    print(f"  Condiciones: {id_eventos}")

    # FASE 5: Épocas — excluyendo canales frontales
    print("=" * 50)
    print("FASE 5: Épocas")
    
    canales_picks = mne.pick_types(
        raw.info, 
        eeg=True, 
        exclude=CANALES_EXCLUIR  # excluir frontales
    )

    epochs = mne.Epochs(
        raw,
        eventos,
        event_id=id_eventos,
        tmin=-1,
        tmax=4,
        picks=canales_picks,
        baseline=(-1, 0),
        reject=None,
        preload=True,
        verbose=False
    )

    print(f"  Épocas totales: {len(epochs)}")
    for cond in epochs.event_id:
        print(f"    {cond}: {len(epochs[cond])}")

    if visualizar:
        epochs.plot(
            n_epochs=5, 
            title='Fase 5 — Épocas segmentadas',
            scalings=dict(eeg=50e-6)
        )
        plt.show()

    return epochs
```

### Función de cálculo de PSD por bandas

Una vez obtenidas las épocas preprocesadas, el siguiente paso del proyecto es extraer las características espectrales de cada época mediante el cálculo de la **densidad espectral de potencia (PSD)** usando el método de Welch.

**¿Qué hace el método de Welch?** Divide cada época en segmentos con solapamiento, calcula la transformada de Fourier de cada segmento y promedia los resultados. Esto produce una estimación más estable y menos ruidosa de la potencia en función de la frecuencia que una FFT directa sobre toda la época.

**¿Qué retorna esta función?** La función `calcular_psd_bandas` recorre cada condición experimental y cada canal disponible, y para cada combinación calcula la potencia integrada en las cinco bandas de frecuencia de interés (Delta, Theta, Mu, Beta, Gamma). El resultado se devuelve como un diccionario anidado: `{condición → canal → banda → potencia}`.

**¿Por qué se procesan todos los canales y no solo C3, C4, Cz?** Aunque la literatura señala estos electrodos como los más relevantes para BCI motor, calcular la PSD en todos los canales permite verificar empíricamente esa hipótesis con los datos del proyecto, en lugar de asumirla a priori. La identificación de canales relevantes se hará después con criterios estadísticos sobre el conjunto completo de sujetos.

```python
from scipy.signal import welch


def calcular_psd_bandas(epochs):
    """
    Calcula la potencia espectral por banda para cada canal y condición.

    Parámetros:
        epochs (mne.Epochs): Épocas preprocesadas

    Retorna:
        dict: Diccionario con potencia por condición, canal y banda
    """
    fs = epochs.info['sfreq']  # frecuencia de muestreo
    canales = epochs.ch_names  # todos los canales disponibles

    resultados = {}

    for condicion in epochs.event_id:
        data = epochs[condicion].get_data()  # shape: (épocas, canales, muestras)
        resultados[condicion] = {}

        for i, canal in enumerate(canales):
            señal_canal = data[:, i, :]  # todas las épocas de este canal

            potencias = {}
            for banda, (fmin, fmax) in BANDAS.items():
                # Calcular PSD con método de Welch vectorizado
                freqs, psd = welch(señal_canal, fs=fs, nperseg=int(fs * 2))

                # Seleccionar frecuencias de la banda
                idx_banda = np.logical_and(freqs >= fmin, freqs <= fmax)

                # Potencia promedio en la banda (promedio sobre épocas)
                potencia_banda = np.mean(psd[:, idx_banda], axis=(0, 1))
                potencias[banda] = potencia_banda

            resultados[condicion][canal] = potencias

    return resultados
```

### Visualización de PSD por canal y condición

Para visualizar los resultados del PSD del sujeto S001, se construye una función que grafica la potencia en una banda específica a lo largo de todos los canales, mostrando una barra por cada condición experimental.

Esta visualización permite identificar de un vistazo:

- **Qué canales** presentan mayor potencia en cada banda.
- **Diferencias visuales** entre condiciones para un mismo canal.
- **Patrones espaciales** de actividad cerebral según la condición.

Se grafican las bandas Mu (8–12 Hz) y Beta (13–30 Hz) por ser las más relevantes para el análisis BCI motor.

```python
def graficar_psd_canales(psd_resultados, banda='Mu', canales_mostrar=None):
    """
    Grafica la potencia en una banda específica para todos los canales
    y condiciones, permitiendo identificar canales significativos.
    """
    condiciones = list(psd_resultados.keys())
    canales = canales_mostrar or list(psd_resultados[condiciones[0]].keys())

    potencias = {cond: [psd_resultados[cond][c][banda] 
                        for c in canales] for cond in condiciones}

    x = np.arange(len(canales))
    ancho = 0.15
    colores = ['#2196F3', '#4CAF50', '#FF9800', '#E91E63', '#9C27B0']

    fig, ax = plt.subplots(figsize=(20, 5))
    for i, (cond, color) in enumerate(zip(condiciones, colores)):
        ax.bar(x + i * ancho, potencias[cond], ancho, 
               label=cond, color=color, alpha=0.8)

    ax.set_xticks(x + ancho * 2)
    ax.set_xticklabels(canales, rotation=90, fontsize=8)
    ax.set_xlabel('Canal')
    ax.set_ylabel('Potencia (µV²/Hz)')
    ax.set_title(f'Potencia en banda {banda} por canal y condición — S001')
    ax.legend()
    plt.tight_layout()
    plt.show()

# Graficar banda Mu y Beta
graficar_psd_canales(psd_s001, banda='Mu')
graficar_psd_canales(psd_s001, banda='Beta')
```

### Rutina general — Procesamiento de todos los sujetos

Una vez validado el flujo de preprocesamiento y extracción de PSD sobre un sujeto de prueba, se construye una rutina que aplica el flujo completo sobre múltiples sujetos del dataset y almacena los resultados en un DataFrame estructurado.

**¿Qué hace la rutina?** Para cada sujeto en la lista, la función:

1. Carga y concatena los runs de interés del sujeto (R03, R04, R07, R08, R11, R12).
2. Aplica el pipeline completo de preprocesamiento (filtrado, CAR, segmentación).
3. Calcula la PSD por bandas en todos los canales válidos.
4. Almacena los resultados como filas en un registro acumulativo.

Cada fila del DataFrame final corresponde a una combinación única de **sujeto × condición × canal**, con columnas para las potencias en cada banda. El bloque `try/except` permite que si un sujeto falla por algún motivo (archivo corrupto, épocas insuficientes), la rutina continúe con el siguiente sin detenerse

```python
def procesar_todos_sujetos(archivos_validos, sujetos, visualizar=False):
    """
    Aplica el flujo completo sobre todos los sujetos y retorna un DataFrame
    con la potencia por banda, canal, condición y sujeto.
    
    Parámetros:
        archivos_validos (list): Lista de rutas de archivos EDF
        sujetos (list): Lista de sujetos a procesar
        visualizar (bool): Si True, grafica las fases de preprocesamiento
    
    Retorna:
        pd.DataFrame: DataFrame con resultados de todos los sujetos
    """
    registros = []

    for sujeto in sujetos:
        print(f"\n{'='*50}")
        print(f"Procesando {sujeto}...")
        print(f"{'='*50}")

        try:
            # Cargar y concatenar runs
            raw = cargar_sujeto(archivos_validos, sujeto)

            # Preprocesar
            epochs = preprocesar_eeg(raw, visualizar=visualizar)

            # Calcular PSD por bandas
            psd = calcular_psd_bandas(epochs)

            # Guardar resultados
            for condicion in psd:
                for canal in psd[condicion]:
                    fila = {
                        'sujeto': sujeto,
                        'condicion': condicion,
                        'canal': canal,
                    }
                    for banda in BANDAS:
                        fila[f'potencia_{banda}'] = psd[condicion][canal][banda]

                    registros.append(fila)

        except Exception as e:
            print(f"  Error procesando {sujeto}: {e}")
            continue

    df = pd.DataFrame(registros)
    return df
```

### Identificación de canales con mayor diferencia entre condiciones

Para seleccionar los canales más relevantes de manera objetiva, se calcula la **varianza entre condiciones** de la potencia promedio para cada canal. Un canal con alta varianza entre condiciones indica que su valor de potencia cambia significativamente según la tarea ejecutada, lo que lo convierte en candidato para diferenciar las condiciones.

**Procedimiento:**
1. Para cada canal y cada condición, se calcula la potencia promedio (sobre sujetos y épocas).
2. Para cada canal, se calcula la varianza entre las potencias promedio de las distintas condiciones.
3. Se ordenan los canales de mayor a menor varianza y se selecciona el top 10 para cada banda de interés (Mu y Beta).

Los canales identificados serán los utilizados en el análisis inferencial posterior, ya que son los que presentan, según los datos, mayor capacidad para diferenciar las condiciones experimentales.

```python
def identificar_canales_relevantes(df, banda, top_n=10):
    """
    Identifica los canales con mayor diferencia de potencia
    entre condiciones en una banda específica.
    """
    # Potencia promedio por canal y condición
    pivot = df.groupby(['canal', 'condicion'])[f'potencia_{banda}'].mean().unstack()
    
    # Varianza entre condiciones por canal
    varianza = pivot.var(axis=1).sort_values(ascending=False)
    
    print(f"\n=== Top {top_n} canales más relevantes — Banda {banda} ===")
    print(varianza.head(top_n))
    
    # Graficar
    fig, ax = plt.subplots(figsize=(12, 5))
    varianza.head(top_n).plot(kind='bar', ax=ax, color='steelblue', alpha=0.8)
    ax.set_title(f'Canales con mayor varianza entre condiciones — Banda {banda}')
    ax.set_xlabel('Canal')
    ax.set_ylabel('Varianza de potencia')
    ax.tick_params(axis='x', rotation=45)
    plt.tight_layout()
    plt.show()
    
    return varianza.head(top_n).index.tolist()


# Calcular canales relevantes en Mu y Beta
canales_mu   = identificar_canales_relevantes(df_resultados, 'Mu',   top_n=10)
canales_beta = identificar_canales_relevantes(df_resultados, 'Beta', top_n=10)

print(f"\nCanales relevantes en Mu:   {canales_mu}")
print(f"Canales relevantes en Beta: {canales_beta}")
```

## Cálculo de ERD/ERS

La Desincronización Relacionada con Eventos (ERD) y la Sincronización 
Relacionada con Eventos (ERS) cuantifican el cambio relativo de potencia 
durante una tarea respecto al período de reposo. Se calculan como:

$$ERD/ERS(\%) = \frac{P_{tarea} - P_{reposo}}{P_{reposo}} \times 100$$

- Valores **negativos** → ERD (desincronización, activación cortical)
- Valores **positivos** → ERS (sincronización, inhibición cortical)

Este cálculo normaliza la potencia respecto al reposo, permitiendo comparar 
entre sujetos y canales de forma más robusta que la potencia absoluta.

```python
def calcular_erd_ers(df):
    """
    Calcula ERD/ERS para cada sujeto, canal y banda,
    usando el reposo como línea base.
    """
    # Potencia de reposo por sujeto y canal
    reposo = df[df['condicion'] == 'reposo'].copy()
    reposo = reposo.rename(columns={
        f'potencia_{b}': f'reposo_{b}' for b in BANDAS
    })
    reposo = reposo[['sujeto', 'canal'] + [f'reposo_{b}' for b in BANDAS]]

    # Merge con el DataFrame completo
    df_erd = df[df['condicion'] != 'reposo'].copy()
    df_erd = df_erd.merge(reposo, on=['sujeto', 'canal'], how='left')

    # Calcular ERD/ERS por banda
    for banda in BANDAS:
        df_erd[f'erd_{banda}'] = (
            (df_erd[f'potencia_{banda}'] - df_erd[f'reposo_{banda}']) /
            df_erd[f'reposo_{banda}']
        ) * 100

    # Eliminar columnas auxiliares
    df_erd = df_erd.drop(columns=[f'reposo_{b}' for b in BANDAS])

    return df_erd
```

## Pruebas de Hipótesis

### Verificación de normalidad — Prueba de Shapiro-Wilk

Antes de seleccionar las pruebas estadísticas, se evalúa si los datos 
de ERD/ERS siguen una distribución normal mediante la prueba de Shapiro-Wilk.

- **H₀:** Los datos siguen una distribución normal
- **H₁:** Los datos no siguen una distribución normal
- **Criterio:** Si p-valor < 0.05 → se rechaza normalidad → prueba no paramétrica

```python
from scipy.stats import shapiro

def verificar_normalidad(df_erd, banda, canales):
    """
    Aplica Shapiro-Wilk por condición y canal.
    """
    print(f"\n=== Prueba de Shapiro-Wilk — Banda {banda} ===\n")
    resultados_norm = []

    for canal in canales:
        for cond in df_erd['condicion'].unique():
            datos = df_erd[
                (df_erd['canal'] == canal) & 
                (df_erd['condicion'] == cond)
            ][f'erd_{banda}'].dropna()

            if len(datos) >= 3:
                stat, p = shapiro(datos)
                normal = p > 0.05
                resultados_norm.append({
                    'canal': canal,
                    'condicion': cond,
                    'estadistico': round(stat, 4),
                    'p_valor': round(p, 4),
                    'normal': normal
                })

    df_norm = pd.DataFrame(resultados_norm)
    
    # Resumen
    pct_normal = df_norm['normal'].mean() * 100
    print(f"Porcentaje de grupos con distribución normal: {pct_normal:.1f}%")
    
    if pct_normal < 50:
        print("→ La mayoría de los grupos NO son normales.")
        print("→ Se utilizará prueba NO PARAMÉTRICA (Kruskal-Wallis)")
    else:
        print("→ La mayoría de los grupos SÍ son normales.")
        print("→ Se utilizará prueba PARAMÉTRICA (ANOVA)")
    
    return df_norm

# Verificar normalidad en canales significativos
df_normalidad_mu   = verificar_normalidad(df_erd, 'Mu', canales_significativos)
df_normalidad_beta = verificar_normalidad(df_erd, 'Beta', canales_significativos)
```

### Hipótesis 3 — Diferencias entre las cuatro condiciones experimentales

Se evalúa si existe diferencia significativa en el ERD/ERS entre las cuatro condiciones experimentales (`mov_real_der`, `mov_real_izq`, `img_der`, `img_izq`) en cada canal y banda de interés.

Dado que cada sujeto aporta un valor de ERD/ERS para cada una de las cuatro condiciones, las muestras son **dependientes** (medidas repetidas). Por ello, la prueba apropiada es **Friedman**, equivalente no paramétrico del ANOVA de medidas repetidas.

- **H₀:** No existe diferencia en el ERD/ERS de las bandas Mu y Beta entre las cuatro condiciones experimentales.
- **H₁:** Al menos una de las cuatro condiciones presenta un ERD/ERS significativamente diferente a las demás.
- **Nivel de significancia:** α = 0.05
- **Corrección por múltiples comparaciones:** Bonferroni, dado que se realiza una prueba por cada canal de interés.

```python
from scipy.stats import friedmanchisquare


def prueba_friedman(df_erd, banda, canales, alpha=0.05):
    """
    Aplica la prueba de Friedman por canal sobre las cuatro condiciones
    experimentales (medidas repetidas dentro de cada sujeto).

    Parámetros:
        df_erd (pd.DataFrame): DataFrame con columnas sujeto, condicion,
                               canal y erd_<banda>.
        banda (str): nombre de la banda ('Mu', 'Beta', etc.).
        canales (list): lista de canales sobre los que aplicar la prueba.
        alpha (float): nivel de significancia.

    Retorna:
        pd.DataFrame con estadístico, p-valor crudo, p-valor corregido
        (Bonferroni) y decisión sobre H0 para cada canal.
    """
    condiciones = ['mov_real_der', 'mov_real_izq', 'img_der', 'img_izq']
    columna = f'erd_{banda}'
    resultados = []

    for canal in canales:
        # Filtrar datos del canal y reorganizar como matriz sujeto x condición
        df_canal = df_erd[df_erd['canal'] == canal]
        pivot = df_canal.pivot_table(
            index='sujeto',
            columns='condicion',
            values=columna
        )

        # Conservar solo sujetos con valor en las 4 condiciones
        pivot = pivot[condiciones].dropna()

        if len(pivot) < 10:
            resultados.append({
                'canal': canal,
                'n_sujetos': len(pivot),
                'estadistico': np.nan,
                'p_valor': np.nan,
                'p_corregido': np.nan,
                'rechaza_H0': False
            })
            continue

        # Aplicar Friedman: cada columna es una condición
        stat, p = friedmanchisquare(*[pivot[c].values for c in condiciones])

        resultados.append({
            'canal': canal,
            'n_sujetos': len(pivot),
            'estadistico': round(stat, 4),
            'p_valor': round(p, 6),
            'p_corregido': np.nan,  # se llena al final
            'rechaza_H0': False
        })

    df_res = pd.DataFrame(resultados)

    # Corrección de Bonferroni
    n_pruebas = df_res['p_valor'].notna().sum()
    df_res['p_corregido'] = (df_res['p_valor'] * n_pruebas).clip(upper=1.0)
    df_res['rechaza_H0']  = df_res['p_corregido'] < alpha

    return df_res


# Aplicar Friedman en bandas Mu y Beta sobre los canales significativos
df_friedman_mu   = prueba_friedman(df_erd, 'Mu',   canales_significativos)
df_friedman_beta = prueba_friedman(df_erd, 'Beta', canales_significativos)

# Mostrar resultados
print("=== Friedman — Banda Mu ===")
print(df_friedman_mu.to_string(index=False))

print("\n=== Friedman — Banda Beta ===")
print(df_friedman_beta.to_string(index=False))

# Resumen
n_sig_mu   = df_friedman_mu['rechaza_H0'].sum()
n_sig_beta = df_friedman_beta['rechaza_H0'].sum()
n_total    = len(canales_significativos)

print(f"\n--- Resumen ---")
print(f"Banda Mu:   {n_sig_mu}/{n_total} canales rechazan H0 (p_corregido < 0.05)")
print(f"Banda Beta: {n_sig_beta}/{n_total} canales rechazan H0 (p_corregido < 0.05)")
```

#### Interpretación de los resultados de Friedman

La prueba de Friedman se aplicó sobre los 19 canales identificados como relevantes en el análisis descriptivo (unión del top-10 en Mu y top-10 en Beta). En cada canal se evaluó si las cuatro condiciones experimentales producen valores de ERD/ERS estadísticamente diferentes. La interpretación se realiza sobre el p-valor corregido por Bonferroni, que ajusta el umbral de significancia para controlar la tasa de error tipo I derivada de las 19 pruebas simultáneas.

**Resultados por banda:**

- **Banda Mu:** 11 de 19 canales rechazan H₀ (57.9%). Los canales significativos son CP4, F4, F6, FC6, P1, P4, P6, P8, PO3, PO4 y POz, con predominio de regiones parietal, parieto-occipital y centro-parietal. El canal con mayor estadístico de Friedman es CP4 (χ² = 64.56, p < 0.001), seguido por P4 (χ² = 54.99) y P6 (χ² = 49.15).

- **Banda Beta:** 13 de 19 canales rechazan H₀ (68.4%). Los canales significativos incluyen AF4, C5, CP4, F4, F5, F6, F7, F8, FC6, P4, P6, P8 y PO4, con mayor presencia de canales frontales laterales además de los parietales. El canal con mayor estadístico es nuevamente CP4 (χ² = 103.03, p < 0.001), seguido por P4 (χ² = 58.71) y F8 (χ² = 44.09).

Dado que Friedman únicamente indica que existe diferencia entre las condiciones sin precisar entre cuáles, en la siguiente sección se aplica el análisis post-hoc de Wilcoxon pareado con corrección de Bonferroni sobre los canales significativos para identificar las parejas de condiciones que efectivamente difieren.

#### Análisis post-hoc — Wilcoxon pareado con corrección de Bonferroni

Una vez confirmado mediante Friedman que existen diferencias significativas entre las cuatro condiciones experimentales en varios canales, se aplica un análisis post-hoc para identificar entre qué pares de condiciones se encuentran las diferencias.

Como las muestras son pareadas (cada sujeto aporta un valor por condición), se utiliza la prueba de **Wilcoxon de rangos con signo** sobre las seis comparaciones posibles entre las cuatro condiciones:

1. mov_real_der vs mov_real_izq
2. mov_real_der vs img_der
3. mov_real_der vs img_izq
4. mov_real_izq vs img_der
5. mov_real_izq vs img_izq
6. img_der vs img_izq

Para controlar la inflación del error tipo I derivada de las múltiples comparaciones, se aplica la **corrección de Bonferroni** dentro de cada canal (multiplicando el p-valor por el número de comparaciones, n=6). El análisis se restringe a los canales donde Friedman rechazó H₀, en concordancia con el procedimiento estándar de pruebas post-hoc.

```python
from scipy.stats import wilcoxon
from itertools import combinations


def posthoc_wilcoxon(df_erd, banda, canales_significativos, alpha=0.05):
    """
    Aplica Wilcoxon pareado entre todas las parejas de condiciones,
    para cada canal donde Friedman rechazó H0.

    Aplica corrección de Bonferroni dentro de cada canal sobre las
    6 comparaciones pareadas entre las 4 condiciones experimentales.

    Parámetros:
        df_erd (pd.DataFrame): DataFrame con sujeto, condicion, canal y erd_<banda>.
        banda (str): nombre de la banda ('Mu', 'Beta').
        canales_significativos (list): canales donde Friedman fue significativo.
        alpha (float): nivel de significancia.

    Retorna:
        pd.DataFrame con las comparaciones pareadas y su decisión sobre H0.
    """
    condiciones = ['mov_real_der', 'mov_real_izq', 'img_der', 'img_izq']
    parejas = list(combinations(condiciones, 2))
    n_comp = len(parejas)
    columna = f'erd_{banda}'
    resultados = []

    for canal in canales_significativos:
        df_canal = df_erd[df_erd['canal'] == canal]
        pivot = df_canal.pivot_table(
            index='sujeto',
            columns='condicion',
            values=columna
        )[condiciones].dropna()

        for cond_a, cond_b in parejas:
            x = pivot[cond_a].values
            y = pivot[cond_b].values

            try:
                stat, p = wilcoxon(x, y)
            except ValueError:
                stat, p = np.nan, np.nan

            resultados.append({
                'canal': canal,
                'comparacion': f'{cond_a} vs {cond_b}',
                'n_sujetos': len(pivot),
                'estadistico': round(stat, 3) if not np.isnan(stat) else np.nan,
                'p_valor': round(p, 6) if not np.isnan(p) else np.nan,
                'p_corregido': round(min(p * n_comp, 1.0), 6) if not np.isnan(p) else np.nan,
                'rechaza_H0': (p * n_comp < alpha) if not np.isnan(p) else False
            })

    return pd.DataFrame(resultados)


# Filtrar canales significativos por banda según Friedman
canales_sig_mu   = df_friedman_mu[df_friedman_mu['rechaza_H0']]['canal'].tolist()
canales_sig_beta = df_friedman_beta[df_friedman_beta['rechaza_H0']]['canal'].tolist()

print(f"Canales significativos en Mu   ({len(canales_sig_mu)}): {canales_sig_mu}")
print(f"Canales significativos en Beta ({len(canales_sig_beta)}): {canales_sig_beta}")

# Aplicar post-hoc
df_posthoc_mu   = posthoc_wilcoxon(df_erd, 'Mu',   canales_sig_mu)
df_posthoc_beta = posthoc_wilcoxon(df_erd, 'Beta', canales_sig_beta)

print(f"\n=== Post-hoc Wilcoxon — Banda Mu ===")
print(df_posthoc_mu.to_string(index=False))

print(f"\n=== Post-hoc Wilcoxon — Banda Beta ===")
print(df_posthoc_beta.to_string(index=False))

# Resumen por comparación
print("\n--- Resumen por comparación (Mu) ---")
print(df_posthoc_mu.groupby('comparacion')['rechaza_H0'].agg(['sum', 'count']))

print("\n--- Resumen por comparación (Beta) ---")
print(df_posthoc_beta.groupby('comparacion')['rechaza_H0'].agg(['sum', 'count']))
```

#### Interpretación de los resultados del post-hoc

El análisis post-hoc se aplicó sobre los 11 canales significativos en banda Mu y los 13 canales significativos en banda Beta identificados por la prueba de Friedman, evaluando las seis comparaciones pareadas posibles entre las cuatro condiciones experimentales. La interpretación se realiza sobre el p-valor corregido por Bonferroni dentro de cada canal (n=6 comparaciones).

**Resultados por comparación en banda Mu:**

| Comparación | Canales que rechazan H₀ |
|---|---|
| mov_real_izq vs img_izq | 11 / 11 |
| mov_real_der vs img_der | 10 / 11 |
| mov_real_izq vs img_der | 9 / 11 |
| mov_real_der vs mov_real_izq | 6 / 11 |
| img_der vs img_izq | 5 / 11 |
| mov_real_der vs img_izq | 4 / 11 |

En la banda Mu las comparaciones más consistentemente significativas son las que contrastan movimiento real con imaginación de la misma mano. La comparación mov_real_izq vs img_izq rechaza H₀ en la totalidad de los canales analizados, y mov_real_der vs img_der lo hace en 10 de 11 canales. Las diferencias entre las dos condiciones de imaginación (img_der vs img_izq) y entre las dos de movimiento real (mov_real_der vs mov_real_izq) son menos consistentes, indicando que en Mu el factor que más diferencia las condiciones es la modalidad (real vs imaginada) más que la lateralidad.

**Resultados por comparación en banda Beta:**

En la banda Beta el patrón es notablemente distinto. ### Hipótesis 1 — Lateralización hemisférica durante movimiento real e imaginación

Esta hipótesis evalúa si existe diferencia significativa en el ERD/ERS entre los electrodos C3 (hemisferio izquierdo) y C4 (hemisferio derecho) durante cada una de las cuatro condiciones experimentales. El principio de control motor contralateral predice que durante el movimiento o imaginación de la mano derecha, el electrodo C3 presenta mayor desincronización (ERD más negativo) que C4, y viceversa para la mano izquierda.

Como cada sujeto aporta un valor de ERD/ERS para C3 y otro para C4 dentro de la misma condición, la comparación es pareada y se aplica la prueba de **Wilcoxon de rangos con signo**. Se evalúan las cuatro condiciones experimentales (`mov_real_der`, `mov_real_izq`, `img_der`, `img_izq`) en las bandas Mu y Beta.

- **H₀:** No existe diferencia en el ERD/ERS entre C3 y C4 dentro de cada condición.
- **H₁:** Existe una diferencia significativa entre C3 y C4 dentro de cada condición, con menor valor en el electrodo contralateral a la mano de la tarea.
- **Nivel de significancia:** α = 0.05
- **Corrección por múltiples comparaciones:** Bonferroni sobre las 4 condiciones evaluadas dentro de cada banda.

```python
def prueba_lateralizacion(df_erd, banda, alpha=0.05):
    """
    Aplica Wilcoxon pareado entre C3 y C4 dentro de cada condición
    experimental, para una banda de frecuencia dada.

    Para cada sujeto se obtiene el ERD/ERS en C3 y en C4, y se compara
    si existe diferencia entre ambos electrodos dentro de la misma
    condición. Se aplica corrección de Bonferroni sobre las 4
    condiciones evaluadas.

    Parámetros:
        df_erd (pd.DataFrame): DataFrame con sujeto, condicion, canal y erd_<banda>.
        banda (str): nombre de la banda ('Mu', 'Beta').
        alpha (float): nivel de significancia.

    Retorna:
        pd.DataFrame con estadístico, p-valor crudo, p-valor corregido,
        decisión sobre H0 y direccionalidad (cuál hemisferio tiene mayor ERD).
    """
    condiciones = ['mov_real_der', 'mov_real_izq', 'img_der', 'img_izq']
    n_pruebas = len(condiciones)
    columna = f'erd_{banda}'
    resultados = []

    for cond in condiciones:
        df_cond = df_erd[df_erd['condicion'] == cond]
        pivot = df_cond.pivot_table(
            index='sujeto',
            columns='canal',
            values=columna
        )

        if 'C3' not in pivot.columns or 'C4' not in pivot.columns:
            resultados.append({
                'condicion': cond,
                'n_sujetos': 0,
                'media_C3': np.nan,
                'media_C4': np.nan,
                'estadistico': np.nan,
                'p_valor': np.nan,
                'p_corregido': np.nan,
                'rechaza_H0': False,
                'hemisferio_mayor_ERD': '—'
            })
            continue

        pivot = pivot[['C3', 'C4']].dropna()

        stat, p = wilcoxon(pivot['C3'].values, pivot['C4'].values)

        media_c3 = pivot['C3'].mean()
        media_c4 = pivot['C4'].mean()

        if media_c3 < media_c4:
            hemisferio = 'C3 (izq)'
        elif media_c4 < media_c3:
            hemisferio = 'C4 (der)'
        else:
            hemisferio = '—'

        resultados.append({
            'condicion': cond,
            'n_sujetos': len(pivot),
            'media_C3': round(media_c3, 3),
            'media_C4': round(media_c4, 3),
            'estadistico': round(stat, 3),
            'p_valor': round(p, 6),
            'p_corregido': round(min(p * n_pruebas, 1.0), 6),
            'rechaza_H0': (p * n_pruebas) < alpha,
            'hemisferio_mayor_ERD': hemisferio
        })

    return pd.DataFrame(resultados)


# Aplicar prueba de lateralización en bandas Mu y Beta
df_lat_mu   = prueba_lateralizacion(df_erd, 'Mu')
df_lat_beta = prueba_lateralizacion(df_erd, 'Beta')

print("=== Hipótesis 1 — Lateralización C3 vs C4 — Banda Mu ===")
print(df_lat_mu.to_string(index=False))

print("\n=== Hipótesis 1 — Lateralización C3 vs C4 — Banda Beta ===")
print(df_lat_beta.to_string(index=False))

# Resumen
print("\n--- Resumen Hipótesis 1 ---")
n_sig_mu   = df_lat_mu['rechaza_H0'].sum()
n_sig_beta = df_lat_beta['rechaza_H0'].sum()
print(f"Banda Mu:   {n_sig_mu}/4 condiciones con diferencia significativa C3 vs C4")
print(f"Banda Beta: {n_sig_beta}/4 condiciones con diferencia significativa C3 vs C4")
```

#### Interpretación de los resultados de Hipótesis 1

La prueba de Wilcoxon de rangos con signo se aplicó comparando los valores de ERD/ERS entre los electrodos C3 y C4 dentro de cada una de las cuatro condiciones experimentales, con corrección de Bonferroni sobre las cuatro pruebas dentro de cada banda.

**Resultados por banda:**

En la banda **Beta**, la totalidad de las condiciones (4 de 4) presentó diferencia significativa entre C3 y C4 tras la corrección. La direccionalidad de las diferencias es consistente en las cuatro condiciones con el principio de control motor contralateral: C3 muestra mayor desincronización (ERD más negativo) que C4 durante las tareas con la mano derecha (mov_real_der: −21.41 vs −14.52; img_der: −16.47 vs −9.05), mientras que C4 presenta mayor desincronización que C3 durante las tareas con la mano izquierda (mov_real_izq: −24.35 vs −16.93; img_izq: −14.96 vs −10.99). En todos los casos el p-valor corregido fue inferior a 0.001.

En la banda **Mu**, 3 de las 4 condiciones presentaron diferencia significativa entre C3 y C4 tras la corrección. Las condiciones mov_real_der (p_corregido = 0.0004), img_der (p_corregido = 0.000008) e img_izq (p_corregido = 0.000011) confirman la lateralización contralateral predicha. La condición mov_real_izq, si bien presentó la direccionalidad correcta (C4: −24.99; C3: −22.08), no alcanzó significancia tras la corrección de Bonferroni (p_corregido = 0.198). Esto puede atribuirse a que la diferencia absoluta entre los hemisferios fue de menor magnitud (~3 unidades) en comparación con las otras condiciones, donde las diferencias rondaron entre 5 y 10 unidades.

**Decisión sobre la hipótesis:**

Se rechaza H₀ en favor de H₁ en 7 de las 8 pruebas realizadas (87.5%), lo que constituye evidencia sólida a favor de la lateralización hemisférica contralateral. La Hipótesis 1 se considera **apoyada por los datos**, especialmente en la banda Beta donde el efecto es completo y consistente. En la banda Mu el efecto es mayoritariamente significativo aunque presenta una excepción en la condición de movimiento real con la mano izquierda.

**Hallazgos relevantes:**

La direccionalidad de las medias es correcta en las 8 comparaciones realizadas, incluso en la única condición que no alcanzó significancia. Esto indica que el principio de control motor contralateral se manifiesta de manera consistente en los datos, aunque la magnitud del efecto varía según la condición y la banda.

Las condiciones de imaginación de movimiento muestran una lateralización tan clara como las de movimiento real, lo cual constituye evidencia de que la representación mental del movimiento activa la corteza motora contralateral de forma medible mediante EEG. Este hallazgo es fundamental para la viabilidad de sistemas BCI basados en imaginación motora, ya que confirma que la firma electrofisiológica que permite discriminar la intención de movimiento se preserva incluso en ausencia de ejecución física.

La banda Beta muestra una lateralización más robusta que la banda Mu en este análisis, complementando el hallazgo del post-hoc previo según el cual Beta es particularmente sensible a la lateralidad de la tarea.

### Hipótesis 2 — Diferencia entre movimiento real e imaginación de movimiento

Esta hipótesis evalúa si la potencia espectral relativa al reposo (ERD/ERS) difiere significativamente entre la ejecución y la imaginación del movimiento, manteniendo constante la mano. La predicción fisiológica es que el movimiento real produce una desincronización cortical más intensa que la imaginación, debido a la activación efectiva del sistema motor.

Las comparaciones pareadas relevantes son `mov_real_der` vs `img_der` (mano derecha) y `mov_real_izq` vs `img_izq` (mano izquierda). Estas comparaciones ya fueron evaluadas con Wilcoxon pareado y corrección de Bonferroni en el análisis post-hoc previo, por lo que en esta sección se extraen, consolidan e interpretan dichos resultados.

- **H₀:** No existe diferencia en el ERD/ERS entre las condiciones de movimiento real e imaginación de movimiento para una misma mano.
- **H₁:** Existe diferencia significativa entre ambas modalidades, con valores de ERD/ERS más negativos durante el movimiento real.
- **Pruebas y corrección:** Wilcoxon pareado con corrección de Bonferroni (ya aplicada en el post-hoc).
- **Criterio de decisión:** se considera la hipótesis apoyada en una banda si la mayoría de los canales significativos rechazan H₀ en las comparaciones de interés.

```python
def extraer_comparaciones_h2(df_posthoc, banda):
    """
    Extrae del DataFrame de post-hoc las comparaciones relevantes para
    la Hipótesis 2: mov_real vs img de la misma mano.

    Parámetros:
        df_posthoc (pd.DataFrame): DataFrame con resultados del post-hoc.
        banda (str): nombre de la banda ('Mu', 'Beta').

    Retorna:
        pd.DataFrame filtrado con las comparaciones relevantes.
    """
    comparaciones_h2 = ['mov_real_der vs img_der', 'mov_real_izq vs img_izq']
    df_h2 = df_posthoc[df_posthoc['comparacion'].isin(comparaciones_h2)].copy()
    df_h2 = df_h2.sort_values(['comparacion', 'canal']).reset_index(drop=True)
    return df_h2


# Extraer comparaciones relevantes para H2 en cada banda
df_h2_mu   = extraer_comparaciones_h2(df_posthoc_mu,   'Mu')
df_h2_beta = extraer_comparaciones_h2(df_posthoc_beta, 'Beta')

print("=== Hipótesis 2 — mov_real vs imaginación — Banda Mu ===")
print(df_h2_mu.to_string(index=False))

print("\n=== Hipótesis 2 — mov_real vs imaginación — Banda Beta ===")
print(df_h2_beta.to_string(index=False))

# Resumen por mano y banda
print("\n--- Resumen Hipótesis 2 ---")
for banda, df_h2 in [('Mu', df_h2_mu), ('Beta', df_h2_beta)]:
    for comp in ['mov_real_der vs img_der', 'mov_real_izq vs img_izq']:
        sub = df_h2[df_h2['comparacion'] == comp]
        n_sig = sub['rechaza_H0'].sum()
        n_tot = len(sub)
        print(f"Banda {banda:5s} | {comp:35s} : {n_sig}/{n_tot} canales rechazan H0")
```

#### Interpretación de los resultados de Hipótesis 2

Las comparaciones pareadas mov_real_der vs img_der y mov_real_izq vs img_izq se extrajeron del análisis post-hoc de Wilcoxon previamente aplicado, considerando los 11 canales significativos en banda Mu y los 13 canales significativos en banda Beta identificados por la prueba de Friedman.

**Resultados por banda:**

En la banda **Mu**, la comparación mov_real_der vs img_der rechazó H₀ en 10 de 11 canales (90.9%) y la comparación mov_real_izq vs img_izq lo hizo en la totalidad de los 11 canales (100%). En conjunto, 21 de las 22 pruebas realizadas resultaron significativas tras la corrección de Bonferroni, lo que constituye evidencia robusta a favor de H₁.

En la banda **Beta**, la comparación mov_real_der vs img_der rechazó H₀ en solo 2 de 13 canales (15.4%) y la comparación mov_real_izq vs img_izq lo hizo en 6 de 13 canales (46.2%). En conjunto, 8 de las 26 pruebas realizadas resultaron significativas (30.8%), lo cual representa evidencia débil a favor de H₁ en esta banda.

**Decisión sobre la hipótesis:**

La Hipótesis 2 se considera **apoyada en la banda Mu** y **no apoyada en la banda Beta**. Este resultado disociado refleja una diferencia fisiológica entre ambas bandas en cuanto a su sensibilidad a la modalidad de la tarea (ejecución frente a imaginación).

**Hallazgos relevantes:**

La banda Mu se comporta como un discriminador efectivo entre movimiento real e imaginación de movimiento, mientras que la banda Beta no lo hace de forma consistente. Esta disociación es coherente con los hallazgos del análisis post-hoc previo, donde se observó que Beta presenta su patrón discriminativo más fuerte en comparaciones de lateralidad (mov_real_der vs mov_real_izq con 13/13 canales significativos), no de modalidad.

Una explicación fisiológica plausible es que la banda Mu refleja el grado de activación de la corteza sensoriomotora, que es máximo durante la ejecución motora real e intermedio durante la imaginación, generando diferencias de potencia detectables. La banda Beta, en cambio, parece sensible principalmente a la organización topográfica del control motor (hemisferio implicado), produciendo patrones similares de ERD/ERS en ejecución e imaginación cuando se compara la misma mano.

Dentro de la banda Beta, el apoyo parcial es asimétrico entre manos: la comparación con la mano izquierda alcanza significancia en 6 de 13 canales, mientras que la comparación con la mano derecha lo hace solo en 2. Una posible explicación es la dominancia manual de la población estudiada, ya que la mayoría de los sujetos son diestros y los patrones de ERD/ERS en la mano dominante tienden a ser más estables y menos discriminables entre modalidades que los de la mano no dominante.

**Implicación para sistemas BCI:**

El hallazgo de que cada banda contribuye con información distinta —Mu para modalidad y Beta para lateralidad— es relevante en el contexto de sistemas BCI. Combinar características espectrales de ambas bandas permitiría discriminar simultáneamente qué tarea está ejecutando el sujeto (movimiento real o imaginación) y con qué mano, mejorando potencialmente la precisión de un clasificador respecto al uso de una sola banda.

## Conclusiones

---

### Síntesis general

El proyecto evaluó la viabilidad de discriminar cuatro condiciones experimentales (movimiento real e imaginación de movimiento, con mano derecha e izquierda) a partir de señales EEG, con vistas a la construcción de sistemas BCI no invasivos. Se procesaron los registros completos de los 109 sujetos del dataset *EEG Motor Movement/Imagery* (OpenNeuro, ds004362), aplicando un pipeline de análisis espectral basado en la densidad espectral de potencia (PSD) calculada por el método de Welch sobre épocas de 5 s, y se cuantificaron los cambios de potencia respecto al reposo mediante ERD/ERS. La evaluación estadística incluyó análisis descriptivo con boxplots, prueba de Friedman para comparación global de las cuatro condiciones, post-hoc de Wilcoxon pareado con corrección de Bonferroni, y pruebas específicas de Wilcoxon pareado para cada hipótesis planteada.

### Hallazgos del análisis exploratorio

El análisis de varianza entre condiciones identificó como canales más discriminativos a los electrodos de la región parietal y centro-parietal (P1, P4, P6, P8, CP4, PO3, PO4, POz) en banda Mu, y a la región frontal lateral y centro-parietal (AF4, F4, F5, F6, F7, F8, CP4) en banda Beta. Es destacable que ninguno de los electrodos motores clásicos C3, C4 o Cz figuró entre los canales con mayor varianza, contrastando con las predicciones más comunes de la literatura BCI orientada al control motor. El canal CP4 presentó el mayor poder discriminativo en ambas bandas, con estadísticos de Friedman de χ² = 64.56 en Mu y χ² = 103.03 en Beta.

El cálculo de ERD/ERS confirmó la presencia de desincronización generalizada durante todas las condiciones de tarea respecto al reposo, con valores típicamente entre −10% y −25%, lo que es consistente con la activación cortical esperada durante movimiento e imaginación. La verificación específica sobre los electrodos motores clásicos C3, C4 y Cz mostró ERD presente en todas las condiciones pero con menor magnitud y consistencia que en los canales parietales identificados estadísticamente.

### Decisiones sobre las hipótesis

**Hipótesis 1 — Lateralización hemisférica contralateral.** La hipótesis se considera apoyada por los datos. La prueba de Wilcoxon pareada entre C3 y C4 dentro de cada condición confirmó la lateralización contralateral en 7 de 8 pruebas (3/4 condiciones en banda Mu y 4/4 en banda Beta). La direccionalidad de las medias fue correcta en las 8 comparaciones, indicando que el principio de control motor contralateral se manifiesta de forma consistente, aunque con magnitudes variables. La banda Beta mostró un efecto más robusto que la banda Mu para esta hipótesis.

**Hipótesis 2 — Diferencia entre movimiento real e imaginación.** La hipótesis se considera apoyada en la banda Mu y no apoyada en la banda Beta. En Mu, 21 de 22 pruebas (95.5%) mostraron diferencias significativas entre las condiciones de movimiento real y de imaginación de la misma mano. En Beta, solo 8 de 26 pruebas (30.8%) alcanzaron significancia. Este resultado disociado refleja que la banda Mu es sensible a la modalidad de la tarea (real frente a imaginada) mientras que la banda Beta es relativamente invariante a esta dimensión.

**Hipótesis 3 — Diferencias globales entre las cuatro condiciones.** La hipótesis se considera apoyada por los datos. La prueba de Friedman rechazó H₀ en 11 de 19 canales en banda Mu (57.9%) y en 13 de 19 canales en banda Beta (68.4%), confirmando que existen diferencias significativas en ERD/ERS entre las cuatro condiciones experimentales en una proporción sustancial de los canales analizados. Nueve canales rechazaron H₀ en ambas bandas simultáneamente (CP4, F4, F6, FC6, P1, P4, P6, P8, PO4), constituyendo el conjunto más robusto para discriminar las condiciones.

### Síntesis interpretativa

Los resultados muestran una **disociación funcional entre las bandas Mu y Beta**: la banda Mu se comporta como discriminadora efectiva de la modalidad de la tarea (ejecución frente a imaginación), mientras que la banda Beta lo hace principalmente respecto a la lateralidad (mano derecha frente a izquierda), particularmente en condiciones de movimiento real donde rechazó H₀ en la totalidad de los canales evaluados. Esta complementariedad funcional sugiere que las dos bandas aportan información independiente sobre la actividad motora cortical, y que un sistema BCI basado en EEG podría beneficiarse de integrar características de ambas bandas en lugar de depender de una sola.

El hallazgo de que los canales parietales y centro-parietales (especialmente CP4) presentaron mayor poder discriminativo que los electrodos motores primarios clásicos abre una línea de discusión relevante. Una posible interpretación es que las tareas del dataset no involucran únicamente la corteza motora primaria sino también los procesos de planificación e integración sensoriomotora propios de la corteza parietal posterior. La variabilidad anatómica entre sujetos y la propagación volumétrica de la actividad eléctrica también pueden contribuir a este resultado.

### Limitaciones

El análisis se realizó sobre la potencia integrada en bandas mediante PSD de Welch, sin considerar la dinámica temporal del ERD/ERS. Una limitación adicional es la exclusión de los nueve canales más contaminados por artefactos sin aplicar técnicas de descomposición como ICA, que podrían haber permitido conservarlos limpios. El rechazo automático de épocas se desactivó tras la exclusión de canales contaminados, lo cual aumenta el N útil pero potencialmente conserva épocas con artefactos residuales. La selección de canales relevantes se basó en varianza entre condiciones promedio sobre todos los sujetos, sin considerar variabilidad inter-sujeto, lo cual podría enmascarar diferencias topográficas individuales relevantes para sistemas BCI personalizados.
